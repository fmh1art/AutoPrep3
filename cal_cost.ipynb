```python
from __future__ import annotations

import json
import os
import glob as _glob
from typing import Any, Dict

def _load_json(path: str) -> Any:
    try:
        with open(path, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        return None
    except Exception as e:
        print(f'[WARN] 读取 JSON 失败: {path} -> {e}')
        return None

def _load_text(path: str) -> str | None:
    try:
        with open(path, 'r', encoding='utf-8') as f:
            return f.read()
    except FileNotFoundError:
        return None
    except Exception as e:
        print(f'[WARN] 读取文本失败: {path} -> {e}')
        return None

def load_one_log(tmp_subdir:str):
    log_root = os.path.join('./_tmp', tmp_subdir, 'log')
    # 得到当前路径下所有文件夹名字
    cases = []
    log_dirs = [d for d in os.listdir(log_root) if os.path.isdir(os.path.join(log_root, d))]
    for uid in log_dirs:
        final_root = os.path.join(log_root, uid, 'log')

        execution_steps = _load_json(os.path.join(final_root, 'execution_steps.json'))
        token_usage = _load_json(os.path.join(final_root, 'token_usage.json'))

        case =  {
            "execution_steps": execution_steps,
            "token_usage": token_usage
        }
        cases.append(case)
    return cases
    
```
